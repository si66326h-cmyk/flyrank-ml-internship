## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

# 1. Ensure we are looking in the right folder within Colab
if os.path.exists('/content/flyrank-ml-internship'):
    %cd /content/flyrank-ml-internship
elif os.path.exists('/content/flyrank-ml-internship-starter'):
    %cd /content/flyrank-ml-internship-starter

# 2. Load the dataset using a clean relative path
try:
    df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
except FileNotFoundError:
    # If Colab's terminal is already sitting inside work/notebooks, we go up two levels
    df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# 3. Apply the standard pipeline filters (pages with traffic & live at least 90 days)
filtered_df = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)].copy()

# 4. Calculate your 2-3 real supporting numbers
total_valid_pages = len(filtered_df)
avg_word_count = filtered_df['word_count'].mean()
down_trend_pct = (filtered_df['trend_direction'] == 'down').mean() * 100

print(f"Total valid content items analyzed: {total_valid_pages:,}")
print(f"Average word count across pages: {avg_word_count:.1f} words")
print(f"Percentage of pages showing a downward trend: {down_trend_pct:.2f}%")
```
